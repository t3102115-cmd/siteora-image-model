## Load base model + Siteora LoRA

```python
from inference.generate import SiteoraImageGenerator

BASE_MODEL = 'stabilityai/stable-diffusion-xl-base-1.0'
LORA_PATH = 'models/siteora-lora-standard'  # or a Drive path

generator = SiteoraImageGenerator(base_model=BASE_MODEL, lora_path=LORA_PATH)
```

## Enter a prompt and generate

```python
result = generator.generate(
    prompt='A premium AI startup office at night, futuristic but professional',
    negative_prompt='blurry, low quality, watermark, text',
    width=1024, height=1024,
    num_inference_steps=30,
    guidance_scale=7.0,
    seed=12345,
    lora_strength=1.0,
)
result.image
```

## Try different LoRA strengths

```python
import matplotlib.pyplot as plt

strengths = [0.5, 0.75, 1.0, 1.25]
fig, axes = plt.subplots(1, len(strengths), figsize=(4*len(strengths), 4))
for ax, s in zip(axes, strengths):
    r = generator.generate(result.prompt, seed=result.seed, lora_strength=s, num_inference_steps=20)
    ax.imshow(r.image); ax.set_title(f'strength={s}'); ax.axis('off')
plt.show()
```
